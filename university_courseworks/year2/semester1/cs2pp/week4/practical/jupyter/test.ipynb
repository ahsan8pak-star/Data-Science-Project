```python
# Define the division function
def myDivide(m, n):
    quotient = m / n
    
    return quotient
```

```python
try:
    m = int(input("Enter the value of m : "))
    n = int(input("Enter the value of n: "))

    assert (abs(m)) >= (abs(n)), "This is an improper fraction -> mixed fraction.\nWe want a standard regular fraction."

    myDivide(m, n)

except ZeroDivisionError:
    print("You MUPPET!\nYou. Can't. DIVIDE. ANYTHING. BY. 0 (ZERO)!!\nIDIOT!!!")

else:
    print(f"fraction: {m} / {n}")

finally:
    print("Thanks for using 'myDivide' function")
```
