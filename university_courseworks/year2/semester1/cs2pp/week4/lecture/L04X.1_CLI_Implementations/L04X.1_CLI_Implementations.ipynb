- https://docs.python.org/3/library/sys.html#sys.argv
- This works as a list of command-line arguments, the first being the name of the Python script.

### `argparse`

`argparse` is also part of the standard library.
- https://docs.python.org/3/library/argparse.html

It provides a relatively easy way to define and parse command-line arguments, offering additional flexibility and user-friendliness in comparison to manual processing.

Features include:
- simplified parsing and automatic validation
- automatic help messages
- type checking
- clear default values
- easy short and long flag options

---
#### Example
The `example_argparse.py` script contains the `argparse` equivalent of the previous example.

**First, explore this script in the editor.**

````python
!python example_argparse.py -h
````

````python
!python example_argparse.py -i infile -o outfile
````

````python
!python example_argparse.py -i -o outfile --threshold 6
````

---
### `sys.argv` vs. `argparse`

`sys.argv` is acceptable for smaller scripts or if you need very custom parsing logic. However, you must **manually handle** each argument, usage messages, error checks, etc.

`argparse` offers **built-in help** generation, type checking, and error handling, which saves time and reduces code complexity.

- Because many Python developers favour readability and maintainability, `argparse` is typically recommended for anything but the simplest cases.  Nonetheless, understanding how to work with `sys.argv` directly is a valuable skill, particularly if you need total control over the parsing process or if working in environments where third-party or even some standard-library solutions may not be available.

- Other options are also available: https://docs.python.org/3/library/optparse.html#choosing-an-argument-parser
