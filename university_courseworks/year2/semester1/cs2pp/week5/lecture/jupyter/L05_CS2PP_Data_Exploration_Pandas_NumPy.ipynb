<h3 style="color:Crimson;">Data Pre-Processing</h3>

- Define and select the amount and kind of data for the full analysis (**scope**).
    - Assure it is **fit for consumption**, meeting the needs of your model and the data consumers.
    - Consistent and unambiguous.
    

- Define a procedure to address issues in data quality (**cleaning**).
    - **Noise**: some modification of the original values (e.g., distortion of a person’s voice during a telephone call)
    
    - **Outliers**: data objects with characteristics that are considerably different from most of the other members of the set
    
    - **Duplicated values**: repeated entries or multiple entries (e.g., 1 person with 2 ages)
    
    - **Missing values**: data was not collected or was lost
        - Eliminate
        - Ignore
        - Estimate
            - Define a default
            - Assign to most common value
            - Assign to the mean or median
            - Interpolate or extrapolate
            - Fit to a curve

- Define how the data might be **transformed** or **reduced**.
    - To make the data better fit the upcoming analyses and machine learning algorithms.
    
        - **Correlated** features might be removed from consideration.
    
        - Data might best be **transformed** to fit the requirements of a model.
    
            - **Normalisiation/Standardisation**: transform the **scale** of the variables to reduce bias or improve processing speed
                - Min-Max (feature scaling)
                    - $$x_{scaled} = \frac{x - x_{min}}{x_{max} - x_{min}}$$
                - Z-score (standard score)
                    - $$x_{scaled} = \frac{x-x_{mean}}{x_{stddev}} = \frac{x-\mu}{\sigma}$$
              

     
- Engineering of **features** to provide the most relevant variables that will enhance the quality of the results.
    - **Ordinal Encoding**: Assign an integer value to a particular label.
        - Good for naturally ordered relationships (e.g., 1st, 2nd, 3rd,…)
    
    - **One-Hot Encoding**: Transform labels into binary array, with 1 denoting a specific label.
        - Good for non-ordinally-related data (e.g., dog, cat, lizard)
    
    - **Dummy Variable Encoding**: Similar to one-hot, but representing C categories with C-1 binary variables.
        - Reduces dimensionality of data.
        - E.g., for 3 colours RGB, (R=0; B=0) implies G=1
    
    - Calculating new feature values
        - Calculate BMI from height and weight data

### `Series`

`Series` in `pandas` are **one dimensional arrays** where each index in the array is labelled. They are constructed using:

<div class="alert alert-block alert-info">
  &emsp;<strong>pandas.Series(<em>data</em>, <em>index</em>)</strong>
</div>

- The `data` argument must be either an array or list, a dictionary, or a scalar value (`int` or `float`).


- If no `index` is given, then one is constructed using integers: 
    - `0, 1, ..., len(data)-1`.

````python
# Basic, unlabelled Series creation:
sr1 = pd.Series([3, 4, 5])
sr1
````

````python
# Basic, labelled Series creation:
sr2 = pd.Series([3, 4, 5], index=['one', 'two', 'three'])
sr2
````

````python
# Labelled Series from dictionary data:
dc = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
sr = pd.Series(dc)
sr
````

---

`Series` in `pandas` are **ordered**.


Values can be accessed both by their `index` *label* and also by a numeric index.

### `DataFrame`

- Data can be stored in **tables** with one column for each **attribute/feature/variable**, and one row for each data **object/item**.

<div class="alert alert-block alert-info">

<h3 style="color:Crimson;">Constructing DataFrames from Series</h3>

We have seen that an isolated column from a `DataFrame` has `Series` type.  


It follows that `DataFrame`s can be built by grouping together `Series`, with each `Series` making up a column of the `DataFrame`.


The indices of the `Series` will be used to construct the

````python
# Construct DataFrame from multiple Series objects
# using random numbers as data
a = pd.Series(np.random.randn(4), 
              index=['one', 'two', 'three', 'four'])
b = pd.Series(np.random.randn(4), 
              index=['one', 'two', 'three', 'four'])
c = pd.Series(np.random.randn(4), 
              index=['one', 'two', 'three', 'four'])

# Construct the DataFrame from a dictionary of Series objects
dfs = pd.DataFrame({"A": a, "B": b, "C": c})
dfs
````

Here, you are actually trying to concatenate a `DataFrame` with a *`Series`* in the latter two cases. 

Pandas is happy enough to let you do this, but the logic it uses is not intuitive.

When you concatenate a `Series` to a `DataFrame`, `pandas` treats the `Series` as a *column*, not as a row. 
- It tacks the Series on as a new column.
- It aligns on the index (which likely does not match anything), so it fills with NaNs.

-----

Use **double brackets** on `.iloc[[]]` or `.loc[[]]` or give a *`.loc[same:same]`* range to indicate that you want a `DataFrame` back.

- The double brackets ensure you get a `DataFrame` with one row, not a `Series`.
- `pandas` interprets `.loc[same:same]` as a *slice*, not a point lookup.


---

### JSON

**JSON** (**J**ava**S**cript **O**bject **N**otation) is a way of representing data structures as text. 

The syntax is very similar to dictionaries and lists
in Python. `pandas` can construct a `DataFrame` from a JSON string or file.


````python
import pandas as pd
from io import StringIO  # Required for json_read, which expects "file-like" input
````

````python
# Create DataFrame from JSON string
json_string = '{"A":[1,2,3], "B":[4,5,6], "C":[7,8,9]}'
j = pd.read_json(StringIO(json_string))
j
````

-----

**Addressing the `FutureWarning`**

In earlier versions of `pandas`, `read_json` could accept string literals.  Eventually, the ambiguity of whether it would be a string literal or a file name became awkward.  The developers implemented a `FutureWarning` to be displayed where string literals were passed:

```python
# FutureWarning: Passing literal json to 'read_json' is deprecated and will be removed in a future version. To read from a literal string, wrap it in a 'StringIO' object.  
  j = pd.read_json(json_string)
```

Implemented as a warning, users could continue to successfully pass string literals, while being informed that a code change would prevent them from doing so in a future `pandas` version release.

````python
from io import StringIO

json_string = '{"A":[1,2,3], "B":[4,5,6], "C":[7,8,9]}'

'''
StringIO is a class that creates an in-memory file-like 
    object that behaves like an open text file.
Future implementations will expect to open a file with the
    first argument, and we are making the string "like a file".
'''

# Wrap the string in StringIO so pandas reads it as a buffer.
j = pd.read_json(StringIO(json_string))

print(j)
````

### `describe()`

- This gives us various statistics of the data:
    - **count** - total number of values in the column
    - **mean** - average of values in the column
    - **std** - standard deviation of values in the column
    - **min** - minimum value in the column
    - **25%** - 1st quartile, 25% of the values are less than this value
    - **50%** - median, 50% of the values are less than this value
    - **75%** - 3rd quartile, 75% of the values are less than this value
    - **max** - maximum value in the column


- Any **missing** values are ignored in the above calculations.


- Note that these statistics only make sense **for numerical variables**, so the columns with other forms of data (i.e., **qualitative**) are omitted.


- Also note that not all numerical values can be meaningfully analysed in this way (e.g., the mean of a numerical ID is not meaningful).

<h2 style="color:Crimson;">Wide and Long Data</h2>

So far, we have been working with mainly **wide** data, organised with:
- columns corresponding to **variables or attributes**
- rows corresponding to the **objects** we are collecting data for
- **values** at each data point
    - Any number of these might be **missing**



---

Another way of organising the same data is to store it as **long** data, where:

- each row stores entries for the **variable name** and **value** columns, and optionally a corresponding **key** column.

**Long data** can be useful where:
- not all variable values are recorded for all objects 
- new/different variables are recorded for different objects (the method of recording might have changed over time).
    - This means that **no missing** values are typically stored in this format, which can be **more efficient**.




-----
Another reason that the difference matters is because **certain tools and methods expect one format or the other.**

-----

We can use **`melt()`** on our **wide** data to give it a **long** format.

Conversely, long-format data can be converted to wide data with **`pivot()`**.

Changing the format can make certain operations simpler and provide proper formatting for certain kinds of plots.

````python
# Re-display for reference
diamonds.head()
````

````python
# This works best when each entry has a unique identifier
diamonds['ID'] = diamonds.index

# All of the original column names become
# entries under the "variable" column.

# Data values are moved to the "data" column.

# ID data is stored in its own column
long = diamonds.melt(id_vars='ID')
long
````

````python
# Convert the long data back to wide format
long.pivot(index='ID', columns='variable', values='value')

# NOTE: Note that pivot() assumes one unique value per combination.
#       If there is duplication, it will raise an error.
````

## `groupby()`

The `groupby()` method groups rows together based on the **values in a specific column**.

<h2 style="color:Crimson;">Reshaping Arrays</h2>

- Any `ndarray` can be **reshaped** to an array of different dimensionality **and the same number of elements** with `ndarray.reshape()`.
    - If possible, the result is a **view** of the original, not a copy.


- A multi-dimensional array can be **collapsed** into one dimension with `ndarray.flatten()`.
    - Returns a **copy**.  A *view* is returned by `ravel()`.
    - The `order` in which these are manipulated is an optional parameter.
        - `C`: C-style, row-major (default), rightmost (last) dimension changes fastest
        - `F`: Fortran-style, column-major, leftmost (first) dimension changes fastest
        - `A`: column-major if Fortran-contiguous in memory
        - `K`: as elements occur in memory


- Transpose dimensions with `ndarray.T`.
    - For 2D, they are flipped.
    - For >2D, the dimension order is reversed.

-----
**Utility**:
- Prepare for linear algebra, machine learning, batch processing, I/O, custom looping...
    - ...because some downstream tool insists on a different structure!

The standard arithmetic operators like `+`,`-`,`*`, and`/` will act element-wise on arrays and can be **broadcast** across a whole array.

Operations with many functions including `np.round`, `np.abs`, `np.log`, `np.exp`, `np.power`, `np.sin`, `np.cos`, etc. can be mapped element-wise simply by applying them to the array.

---
When operating on **two arrays**, `numpy` compares their shapes element-wise. It starts with **trailing dimension** and works its way left. Two dimensions are **compatible** when
- they are of equal size, or

- one of them is size 1.

If these conditions are not met, a `ValueError: operands could not be broadcast together` exception is thrown, indicating that the arrays have incompatible shape.

````python
# Multiply arrays of the same dimension and shape
a = np.arange(1, 10).reshape(3, 3)
b = np.flip(a)  ## np.arange(9, 0, -1).reshape(3, 3)
c = a * b
print(a)
print(b)
print(c)
````

````python
# Adding a vector to a matrix
#   add vector to each row
a = np.arange(1, 21).reshape(5, 4)
b = np.arange(1000, 4001, 1000) # len(b) = 4
c = a + b
c
````

````python
# Adding a vector to a matrix (incompatible dimensions)
#   add vector to each column
a = np.arange(1, 21).reshape(4, 5)
b = np.arange(1000, 4001, 1000) # len(b) = 4
c = a + b
c
````

````python
b[:, np.newaxis].shape
````

````python
# Adding a vector to a matrix (incompatible dimensions, realigned)
a = np.arange(1, 21).reshape(4, 5)
b = np.arange(1000, 4001, 1000) # len(b) = 4
c = a + b[:, np.newaxis]
c
````

**Third-party utilities like Pandas, NumPy, Matplotlib, Seaborn, and scikit-learn let you move data around quickly, but understanding what you are doing, why you are doing it, and what assumptions you are imposing will always matter more than the code itself.**

1. Tools don't replace thinking.
    - Speed and convenience can encourage carelessness.
2. Know what your operations actually mean.
    - `groupby`, `resample`, `rolling` make assumptions about the data and analysis intent.
    - If you cannot explain the line of code, then you are not yet in control of it.
3. Data cleaning is not neutral.
    - Every action changes the story (and sometimes the interesting bits are in the messy spaces!).
4. Be suspicious of defaults.
    - They just promise functionality, not usefulness.
5. Remember to consider scale, uncertainty, and physical meaning.
    - Pandas will happily average quantities that should not be averaged, correlate nonsense with nonsense, and merge incomparable datasets.
    - Your domain knowledge is the only safeguard against confident, but meaningless, results.
