### Python Modular Programming



- A method of breaking large coding tasks into smaller and more manageable subtasks.

- **Variable** - a name bound to an object in memory, referencing information or references to other objects.

- **Function/Method** - repeatably executable named block of coded instructions, possibly with inputs/outputs.

- **Class** - named encapsulation of related data variables and functions.


- **Module** - a bunch of related code saved in a file with the extension `.py`.
    - `import`able namespace, potentially installable


- **Package** - a directory containing of modules/subpackages. (May have `__init__.py` for setup operations on import.)
    - `import`able namespace, potentially installable 


- **Library** - often used interchangeably with "Package"
    - an umbrella term referring to a reusable chunk of code, a **conceptual** collection of packages and/or modules...perhaps just one of these.
    - Often, this is some code that we can `import` and use.


- **Framework** - describes the basic structured flow and architecture of the application using combinations of the above components, often to assemble applications.


- **Environment** - the collection of code elements that are *visible* to your instance of Python.
    - includes basic Python runtime utilities, additional installed elements, and wider system resources.

#### Running Python

- **Python** is just a *language specification*: syntax, semantics, and the standard library.
    - It does **NOT** determine machine-level implementation; **multiple** implementations exist.
 
- **CPython** is the *reference implementation* of Python, written in C, available from python.org.
    - Other implementations exist (e.g., PyPy, Jython, IronPython, MicroPython)
    - CPython *itself* is a compiled C program.

- When Python runs a `.py` file, it is first **parsed** into and abstract syntax tree (AST) and checked for syntax, then compiled into a **bytecode** instruction set, and a cached version is saved (e.g., `__pychache__/*.pyc`).
    - The bytecode is platform agnostic, but specific to CPython version and optimisation level.
    - This process is largely **invisible** to the user/programmer.
    - Caching (e.g., of imported modules) speeds up execution in subsequent executions.
    - The `__main__` code is compiled in memory, not cached.


- From the user perspective, there is no separation between compiling and running code.
    - You can simply write a program and execute it. 
    - You can even execute strings containing Python code within a Python program!
    

---
#### The Python Virtual Machine

The Python Virtual Machine (PVM) is the **interpreter** that executes byte code by looping *through* the individual bytecode version of the Python instructions: fetch–decode–execute.
- This actually calls *pre-compiled native C utilities* to make the original Python code finally touch metal.

- Find a **data** source, plan access, and understand its basic characteristics.

<h2 style="color:Crimson;">Indentation and Comments</h2>

**In Python, indentation acts as *blocks* do in `C` or `Java`:**

C/Java:


Python:

```python
    if x < y:
        z = x
    else:

        else:
            z = y
```

- Block **indentation** is *typically* **4 spaces**.
    - With Jupyter notebooks, in code and markdown cells, the `Tab` key should automatically provide 4 spaces.

- Sometimes, the lack of explicit **block termination** markers can make it difficult to determine which parts of the code belong to which block.
    - As a work-around, you might see a **comment** placed at the end of the block, with matching indentation.

<h2 style="color:Crimson;">Variables</h2>

- **Variables** are names used to refer to stored object values.



- A variable is created the moment you first **assign** a value to it.
    - Only after this can it be used.



- Use the **assignment operator**, `=` to assign a value to a variable.



- Variables **DO NOT** need to be declared with any particular type.
    - In fact, variables can even change type at any time.

    

- Use the intrinsic `type()` function to inspect the data type of a variable.



- **All** variables are implemented as **objects**.
    - They each have associated **methods** and **attributes** that can be accessed using "**dot notation**".

````python
# Use the variable `number` to store the value 1,000,000,000
number = 1_000_000_000    # Using underscores as "numeric separators"
number = 10_00_00_0000    # ... you can put them wherever you like!

# Display its type
type(number)
````

````python
# Assign a different type of value to the `number` variable
number = -0.0625

type(number)
````

<h3 style="color:Crimson;">Variable Names</h3>

A variable name can be **short** (like `x` and `y`) or  more **descriptive** (e.g., `age`, `carname`, `total_volume`). 


- **Rules** for Python variables:
    - A variable name **must** start with a letter or the underscore character.
        - Meaning that a variable name **cannot** start with a number.
    - A variable name can only **contain** alpha-numeric characters and underscores (**A-z, 0-9, and _** ).
    - Variable names are **case-sensitive** (age, Age and AGE are three different variables).

````python
# Some legal variable names
myname = 'Python'
my_name = 'Python'
_my_name = 'Python'
myName = 'Python'
MYNAME = 'Python'
myname2 = 'Python'
````

````python
# Some illegal variable names
#    Running this cell will yield errors.

9myname = 'Python'
my-name = 'Python'
my name = 'Python'
````

````python
# Multi-word variable naming conventions
myVariableName = 'Python'     # Camel Case:  each word, except the first, 
                              #              starts with a capital letter.
MyVariableName = 'Python'     # Pascal Case: each word starts with a capital 
                              #              letter.
my_variable_name = 'Python'   # Snake Case:  each word is separated by an 
                              #              underscore character.
````

````python
# Combining print() with the `.format()` method.

# NOTE: You can break a line in Python without special notation.
#       In some cases, a formal `\` is required.
#       Indentation of the continued line is inconsequential.

print('My name is {} {}, and I have {} years of Python experience.'.format(
    first, last, n))

print('My name is {one} {two}, and I have {three} years of Python experience.'.
      format(one=first, two=last, three=n))

print('My name is', first + ' ' + last, ', and I have', n,
      'years of Python experience.')
````

<h2 style="color:Crimson;">Examples of Future Concepts</h2>


````python
# Dynamically create a new class named 'MyDynamicClass'

#   The new class inherits from 'object' (the base class for all new-style classes in Python)
#   A dictionary defines the attributes and methods for the class; here we define a method 'greet'
#   `type()` acts as a class constructor.

MyDynamicClass = type(
    'MyDynamicClass',          # Name of the new class
    (object,),                 # Tuple of base classes (inheriting from object)
    {
        'greet': lambda self: "Hello from dynamically created class!"  # Define 'greet' method using a lambda function dictionary entry
    }
)

# Create an instance of the dynamically generated class
instance = MyDynamicClass()

# Call the 'greet' method on the instance and print the result
print(instance.greet())  # Expected output: "Hello from dynamically created class!"
````

````python
# Python functions can inspect and even modify their own behaviour during runtime.

def foo():
    # Original function that returns a fixed string
    return "Original Function"


# Print the output of the original function
print("Before:", foo())  # Expected output: "Original Function"
````

````python
# Replace the code object of 'foo' with that of a new lambda function
# The lambda function, when called, returns "Modified Function"
foo.__code__ = (lambda: "Modified Function").__code__


# Print the output after modification
print("After:", foo())   # Expected output: "Modified Function"
````

````python
# Let's go after a specific value in a function...

import dis

def banner():
    return "ORIGINAL"

# First run: original behaviour
print("First call:", banner())
````

````python
# Disassemble the function into its instruction components
print("\nDisassembly of original function:")
dis.dis(banner)
````

````python
# Inspect the constants table of the code object
orig_consts = list(banner.__code__.co_consts)
print("\nOriginal co_consts:", orig_consts)
````

````python
# Patch: replace "ORIGINAL" constant with "PATCHED"
patched_consts = tuple(
    "PATCHED" if c == "ORIGINAL" else c
    for c in orig_consts
)

# Create a new code object by replacing just the constants
new_code = banner.__code__.replace(co_consts=patched_consts)

# Swap same function object, new constants
banner.__code__ = new_code

# Second run: behaviour has changed without redefining
print("\nSecond call (after patch):", banner())
````

````python
# Disassemble, showing that bytecode is the same; only constants differ.

print("\nDisassembly of patched function:")
dis.dis(banner)
````
