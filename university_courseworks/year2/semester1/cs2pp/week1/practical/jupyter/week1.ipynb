# CS2PP Week 1 Practical — Python Fundamentals

This notebook covers the core concepts from Week 1: variables, data types,
arithmetic, strings, conditionals and loops. Each cell can be run independently.


## 1. Variables and Data Types

Python infers types dynamically. Assign a value with `=` and inspect it with `type()`.


```python
name = "A.I.M"
student_id = 34001219
gpa = 3.7
on_course = True

print(name, type(name))
print(student_id, type(student_id))
print(gpa, type(gpa))
print(on_course, type(on_course))
```

## 2. Arithmetic Operations

Use `+`, `-`, `*`, `/`, `//` (floor division), `%` (modulo), `**` (power).


```python
a = 17
b = 5

print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}")
print(f"{a} // {b} = {a // b}")
print(f"{a} % {b} = {a % b}")
print(f"{a} ** {b} = {a ** b}")
```

## 3. String Manipulation

Strings are immutable sequences. Use indexing, slicing and f-strings.


```python
greeting = "Hello, student"

print(greeting[0])        # first character
print(greeting[-1])       # last character
print(greeting[:5])       # first five characters
print(f"{greeting}, {name}!")  # f-string interpolation
print(len(greeting))    # length
```

## 4. Conditionals

Use `if` / `elif` / `else` for branching. Python uses indentation, not braces.


```python
score = 78

if score >= 70:
    grade = "Distinction"
elif score >= 60:
    grade = "Merit"
elif score >= 50:
    grade = "Pass"
else:
    grade = "Fail"

print(f"Score {score} -> {grade}")
```

## 5. Loops

Use `for` to iterate over sequences and `range()` for numeric loops.


```python
print("Counting to 5:")
for i in range(1, 6):
    print(i)

print("
Digits of the student ID:")
for digit in str(student_id):
    print(digit)
```

## Summary

- Variables: assign with `=`, inspect with `type()`
- Arithmetic: `+`, `-`, `*`, `/`, `//`, `%`, `**`
- Strings: index/slice with `[]`, interpolate with f-strings
- Conditionals: `if` / `elif` / `else` with indentation
- Loops: `for` over sequences or `range()`
