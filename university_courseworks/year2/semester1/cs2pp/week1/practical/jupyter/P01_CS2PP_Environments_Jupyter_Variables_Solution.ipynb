## JupyterLab

**Recall** from the lecture: 

- We will be using a notebook system called `Jupyter`. 

- **Reference:** https://jupyterlab.readthedocs.io/en/latest/
    - Also, within the JupyterLab space, `Help` --> `JupyterLab Reference`

- Jupyter notebooks allow you to write **cells of code** and **display their output** as text, images, or interactive widgets, as well as **formatted text descriptions**.
    - Excels in reproducibility
    - Easy to share and publish on-line
    
    
- `Jupyter` runs as a **locally hosted web-app** that you interact with through a web browser. 


- `Jupyter` manages running a language-specific [**kernel**](https://jupyter-notebook.readthedocs.io/en/stable/examples/Notebook/What%20is%20the%20Jupyter%20Notebook.html#Kernels) that executes your code and maintains a global variable state. This means that:
    - Variable and function definitions persist between cells.
    - Cells can be run in **any order**.
    - Running a cell will make a **persistent change** to the global state of your interactive session.


- The **kernel**'s properties are defined by specific code environments.
    - In a Python context, the kernel enables additional local processing before passing code on to CPython for parsing/compiling and PVM, collecting and displaying the results.


- Notebooks are an excellent space to test out ideas with **rapid iteration**.
    - This is highly compatible with **learning** a programming language through experimentation and **data science** workflows.
    - No need to execute entire programs repeatedly.
    - But you can, if you want to.
    - You can also **convert** notebooks into Python scripts when you are happy with the code.
    

**As best practice, write your notebooks to be run sequentially from top to bottom!**

## Stopping the Kernel


**Let's cause a problem!**

The cell below has a very simple, slowly incrementing infinite loop.  When you execute the cell, you will see, to the left of the cell in the square brackets: `[*]`.  An **asterisk** means that the cell is either still running or queued to be run (the latter is the obvious case when there are many of these in the notebook).  If the topmost cell has this asterisk status, it is running; if it is further below in a sequence of asterisks, it is queued.  (We can execute many cells at once by selecting versions of `Run all cells` from the menu bar; see below.)

## Types of Cells

    - Markdown is a plain text document markup language, presenting nicely formatted text with simple annotations in standard plain text.
    - You can also insert HTML, links, tables, code blocks, LaTeX equations, and images.
    - Notebook files can be exported to multiple formats, like PDFs, for greater portability.
    - <span style="color:Crimson;"><b>The text-based descriptions and explanations that your provide as part of your assessment will be written in this format!</b></span>

### Difficulty Loading Jupyter Notebooks

Some students have experienced difficulty launching JupyterLab on their personal machines. 

Launching from an Anaconda prompt (command: `jupyter lab`) or by selecting Jupyter Notebook from the Windows Start menu failed to **automatically launch** the Notebook service in a new browser tab.  Instead, the two options will **present text in a terminal window**, enabling the user to copy and paste a given URL (with token) into the browser; doing this does successfully initiate an active session but is a bit cumbersome. 

For example,

```bash
...
[I 2024-10-30 17:36:28.176 ServerApp] http://localhost:8889/lab?token=5ed4df1e0fbdec96365bc71c6cd3f769dc07d97d7828d8bc
[I 2024-10-30 17:36:28.177 ServerApp] http://127.0.0.1:8889/lab?token=5ed4df1e0fbdec96365bc71c6cd3f769dc07d97d7828d8bc
...
```

Researching this issue in the past, most proposed solutions match our working solution (copy/paste URL from terminal).  The reason for this problem might be related to the use of a local system username that includes unicode characters (e.g., Cyrillic or Chinese characters) that are not properly handled by the system interaction with the launching service.  Other possible issues relate to errors in installation of, or updates to, Anaconda and a variety of potential conflicting system path configurations. 

We are unlikely to provide a universal solution to this problem.  Clever Googling/LLMing might be the remedy of choice, as you can search more specifically to your own context.

### About Ports

Notebooks are run through **servers** on a local port.  Typically, they initially use port `8888` but will try others (typically, sequentially increasing values) if `8888` is in use.  By default, a notebook server runs locally at `127.0.0.1:8888` and is accessible only from `localhost`.  You may access the notebook server from the browser using http://127.0.0.1:8888 directly. 

When launching from the Navigator, port information is found in the address.

In a prompt or terminal:

- List active jupyter sessions with their tokens: 
```
jupyter server list
```

- Launch with a specific port number:
```
jupyter lab --port <port_number>
```

- Kill active jupyter sessions by specifying the port number: 
```
jupyter server stop <port_number>
```

### `conda` vs `pip`

`conda` and `pip` are functionally similar, but certainly not identical, utilities to manage packages of code.  Often, they have overlapping functionality and can be used in tandem, though this can lead to problems.

**Both help us install new tools!**

- `conda`: a cross platform package and environment manager that installs and manages conda packages from the Anaconda repository as well as from the Anaconda cloud and other "channels".
    - Binary installation, no need to compile
    - Manages packages from **other languages** (C, C++, R, etc.)
    - Built-in support for separate environments
    - Robust dependency resolution
    - Over **1,500 packages** on the repository, plus thousands more on the cloud
    - Often requires **wildly unreasonable wait times** to create environments, reconcile version conflicts, and install packages.
    
    
- `pip`: the Python Packaging Authority’s recommended tool for installing packages from the Python Package Index, [PyPI](https://pypi.org/).
    - Complete functionality might require further compilation of non-Python utilities (need appropriate compilers/libraries)
    - Often requires use of other (easily installable) tools to manage environments (e.g., conda and uv)
    - Simple dependency resolution (this has gotten better, but still might lead to subtly broken package combinations)
    - Over **150,000 packages** through PyPI
        - These are not thoroughly inspected; use at your own risk ([EXAMPLE](https://digital.nhs.uk/cyber-alerts/2026/cc-4781)).
    - Typically, installations are **much quicker** than `conda` (sometimes because of less stringent dependency resolution).

### Creating a new `conda` environment

One example of a potential conflict with our `base` environment arises with the Tensorflow library, which is used in machine learning applications.  In particular, to harness the power of the NVIDIA GPUs on these lab PCs, the library needs to be installed alongside a particular set of dependencies.  The `base` environment only supports CPU computations, and Windows machines have a problem with running GPU enabled Tensorflow in the newest versions of the library (beyond 2.10).  Because this is known to work with an older version of Python, we can specify this as part of the new environment, too.

To account for these issues, we can create a new environment for GPU-enabled Tensorflow.

<div class="alert alert-warning" role="alert">
  <strong>Note:</strong> Following the command sequence below will take up <strong>4-5 GB</strong> of storage space. Only do this if you are particularly interested in using GPU-enabled TensorFlow.
  <p>To create a lighter environment, you can <strong>exclude</strong> the following command from the sequence:</p>

</div>

**Open a new Anaconda Prompt, providing access to the `conda` command.**

Enter the following commands, confirming `y` when prompted.  Some of these will take a minute to complete. 
```bash
conda create --name c39tf python=3.9 
conda activate c39tf 
conda install -c conda-forge cudatoolkit=11.2 cudnn=8.1.0 
python -m pip install --upgrade pip 
pip install "numpy<2.0" "tensorflow<2.11" 
```

In order, the preceding commands will:
1. Create the new environment, linked to Python 3.9.
2. Activate the new environment (enter its space).
3. **(Optionally)** Install NVIDIA support tools in the new environment.
4. Install `pip` in the new environment, so that it is consistent with Python 3.9.
5. **Takes a minute**.  Install `tensorflow` utilities in the new environment.

Because this module relies almost exclusively on Python-based applications, there is a **lightweight alternative: `uv`**(https://docs.astral.sh/uv/).  `uv` is an extremely fast Python package and project manager, written in Rust, heavily dependent on `pip`.  We can make `uv` environments very similar to those from `conda`, and we provide tools for you to configure an environment matching the course's Anaconda `base` environment.

Do you already have conda/anaconda?  If you would like to keep using that system and avoid the potential conflicts of two separate installations:
1. Consider whether you would be comfortable uninstalling the existing version and starting anew.
2. Consider creation of a new conda environment that sits within your existing distribution.

### Setting up with `uv`

Documentation: https://docs.astral.sh/uv/

The main advantage here is going to be faster and lightweight portability (but not necessarily lightweight installation...).  The necessary files and commands are tiny and easy to move around (you can easily keep a copy on the `X:` drive to install as needed).  The Anaconda Navigator and all of the related components will not be installed, though.

Details will vary depending on operating system.  These notes will show how this can be done on a Lab PC (**highly recommended that you try this!**).

#### CLU Utility Installation

Visit https://docs.astral.sh/uv/getting-started/installation/ to select the **Standalone installer** command suitable for your system.  The version is not particularly important; go with the latest.

For Lab PCs, 

1. **open a `cmd` prompt** and execute:
```bash
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```
2. Close the prompt and restart `cmd` to add the `uv` executable to your `PATH`.
   
#### Frameworks

There are 2 implementation options:
1. a **project**, which is primarily about keeping code and the associated environment together in the same directory (a bespoke environment for a specific application)
2. a virtual environment that works much like a `conda` environment

-----

#### Project

Projects need a space.

**Extract** the supplied `pp_plocal.zip` (on Blackboard) and **enter** the resulting directory, containing:
- `.python_version`
- `pyproject.toml`
- `uv.lock`

Execute (this creates the project environment as defined by the files):
```bash
uv sync --locked
```
Packages will install in about a minute.  They will be stored under `.venv`.  Adding a `.gitignore`:
```bash
# Python-generated files
__pycache__/
*.py[oc]
build/
dist/
wheels/
*.egg-info

# Virtual environments
.venv
```
makes this space very suitable for storing the environment configuration alongside the code in git.

**Accessing utilities**:
- interactive python (**reference `uv`** to ensure the connection to the project environment):
```bash
C:\Users\uu915119\pp_plocal> uv run python
```

- launch notebook (from this `pp_plocal` directory):
```bash
C:\Users\uu915119\pp_plocal> uv run jupyter lab
```

**Directly access the `uv` environment and use it like a `conda` environment**:
- activate the environment
```bash
C:\Users\uu915119\pp_plocal> .venv\Scripts\activate
```
- Note **new** environment prompt prefix:
```bash
(pp-project) C:\Users\uu915119\pp_plocal>
```
- launch notebook (from this ANY directory, e.g., `X:`):
```bash
(pp-project) X:\> jupyter lab
```
- Exit environment:
```bash
(pp-project) C:\Users\uu915119> deactivate
C:\Users\uu915119>
```

#### Standalone `uv` venv

These instructions assume you have already installed `uv`, as above.  Without using a project space, create a directory for the environment to be stored.
- Configuration takes a bit longer than using the project, because it needs to solve the environment requests, whereas this is done in advance in the project.  This process was used to define the files defining the above project.

Open new `cmd` window and make a directory:
```bash
mkdir pp_elocal
cd pp_elocal
```

Create a `uv` venv called `elocal` with our intended version of Python:
```bash
uv venv elocal --python 3.13.9
```

Activate the env:
```bash
elocal\Scripts\activate
```

Copy over (or point to) `requirements.uv.txt` (on Blackboard)
- NOTE: this is a **different** requirements file than that used in the Anaconda installation!
install the python packages:
```bash
(elocal) C:\Users\uu915119\pp_elocal>uv pip install -r requirements.uv.txt
```


Invoke `python` and `jupyter lab` directly, given the `(elocal)` prefix showing the active environment, much like the direct access of the project environment above.

### Google Colab

**As an absolute last resort...**

Google Colab operates in a Notebook environment that is somewhat different to our implementation in terms of software versions and basic functionality.

If you choose to work with Google Colab, please note that some features in the Jupyter Notebook material prepared for this module will likely not render properly due to some use of HTML and embedded images.  Use of HTML is configured differently in the Google Colab version of Jupyter Notebooks, and producing additional compatible versions would require significant additional effort in preparation and maintenance.  Google Colab also uses a different file system structure that would force similar changes to render linked images.  That said, most of the **examples of code** in the course materials should work there, as Google Colab is currently running **Python 3.13.15**, but this can change without notice.

- You will often need to access datasets that we provide.  These tend to be packaged in archive files with notebooks that used the associated paths.  In the case of Google Colab, you will need to modify the provided paths and arrange **your own data storage solution**:
    - https://colab.research.google.com/notebooks/io.ipynb
    

- To access, log in to a Google account, and go to: https://colab.research.google.com/notebooks/intro.ipynb to get started.
