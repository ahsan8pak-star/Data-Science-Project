````python
""" *args """

# Unnamed, positional arguments
l_args = ['first', 'second', 'third']

def order(one, two, three):
    print(one, two, three)

# Evaluate l_args as a set of positional arguments
order(*l_args)
````

````python
""" **kwargs """

# Named, keyword arguments
d_args = {'four': 4,
        'five': 'five',
        'six': 6.0}

def order(one, two, three, four=None, five=None, six=None):
    print(one, two, three, four, five, six)

order(*l_args, **d_args)
````

````python
parg1 = 'parg1'
l_args = ['second', 'third']
d_args = {'four': 4,
        'five': 'five'}

order(*l_args, parg1, **d_args )
# Note the use of default values for unused kwargs
````

````python
order(*l_args, parg1, six='sixth', **d_args)
````

````python
def arg_test(arg1, *argvals, kwarg1=None, **kwargpairs):
    print("First argument:", arg1)
    
    for anc in argvals:
        print("Another argument from *argvals:", anc)

    print('\nSpecified keyword:', kwarg1)

    for key, value in kwargpairs.items():
        print(f"Found keyword arg: '{key}': {value}")

kws = {'kwarg2': 22,
       'kwarg3': 33,
       'kwarg4': 44,
       'kwarg5': 55}

arg_test('name', 100, 200, 300, 'positional?', **kws)
````

````python
# Pass by value: immutable (e.g., string)

def duplicate(in_string):
    in_string = in_string * 2
    return in_string

x = '! Some message !'
print(f'Input Before: \n\t{x=}')

y = duplicate(x)

print(f'Input After: \n\t{x=}')
print(f'Output: \n\t{y=}')

# The original input variable is unchanged.
````

````python
# Pass by reference: mutable (e.g., list)

def swap(a):
    a[0], a[-1] = a[-1], a[0]
    return a

x = [0, 1, 2, 3, 4, 5, 6]
print(f'Input Before: \n\t{x=}')

y = swap(x)

print(f'Input After: \n\t{x=}')
print(f'Output: \n\t{y=}')

# The original input variable is CHANGED.
````

# Object-Oriented Programming

Object-Oriented programming (OOP) is a computer programming model that organizes software design around data, or objects.

The structure, or building blocks, of object-oriented programming include:

- Classes: user-defined data types that act as the blueprint for individual objects, attributes and methods.

- Attributes: defined within the class template and represent the state of an object. Objects will have data stored in the attribute fields.

- Methods: functions that are defined inside a class that describe the behaviors of an object.

- Objects: instances of a class created with specifically defined data

```python
class className:
    attribute = ...

    def method():
        ...

instance_object = className()
```

````python
# Define a class with instance variables
class Student:
    
    """
    Descriptive formal docstring...
    Contains...methods...
    """
    
    # Class attribute
    count = 0

    # Constructor method
    def __init__(self, name='Default', mark=0):
        # Initialise instance attributes
        self.name = name
        self.mark = mark

        # A "private" attribute
        self.__secret = 'HUSH'
        
        # Access the class attribute
        Student.count += 1
        self.sid = Student.count
        
        # Initialise empty dictionary
        self.modules = {}
    
    # Method to print information about class
    def info(self):
        print(f'{self.name}, {self.sid}: {self.mark}, {self.__secret}')
    
    # String representation
    def __str__(self):
        return self.name
    
    # Object representation
    def __repr__(self):
        return f'{type(self).__qualname__}("{self.name}",{self.mark})'
    
    # An instance method
    def add_submark(self, module, mark):

        if module in self.modules:
            self.modules[module].append(mark)
        
        else:
            self.modules[module] = [mark]
        
        self.__update_mark()

    def __update_mark(self):
        total = 0
        entries = 0
        for key, value in self.modules.items():
            for v in value:
                total += v
                entries += 1

                self.mark = total / entries

    # Polymorphism: operator overloading
    def __add__(self, other):
        return f'{self.name} and {other.name}'
    
    # Polymorphism: allow subclasses to define behaviour
    def classification(self):
        raise NotImplementedError("Subclass must implement abstract method.")

````

````python
# A child class which inherits attributes name, mark and functions
# from the parent class and builds on them
class UGStudent(Student):

    # Constructor
    def __init__(self, name, mark, contact): # now positional, required parameters
        super().__init__(name, mark) # run the constructor of the parent class
        self.emergency_contact = contact

    # Method to print information about class
    def info(self):
        print(f'{self.name}, {self.sid}: {self.mark}. Contact: {self.emergency_contact}')

us1 = UGStudent('Adam', 75, 'Mother')
us1.info()
````

````python
# A variation on previous:
# .info() method relies on parent functionality, then adds functionality

# A child class which inherits attributes name, mark and functions
# from the parent class and builds on them
class UGStudent(Student):

    # Constructor
    def __init__(self, name, mark, contact):
        super().__init__(name, mark) # run the constructor of the parent class
        self.emergency_contact = contact
    
    # Method to print information about class
    def info(self):
        super().info()
        print(f'Contact: {self.emergency_contact}')

us1 = UGStudent('Adam', 75, 'Mother')
us1.info()
````

````python
# Duck typing in class method

class fruit:

    def __init__(self, type):
        self.type = type

    def info(self):
        print(f"Fruit Type: {self.type}")

    def calculate(self, values):
        return(sum(values))
````

````python
class vegetable:

    def __init__(self, type):
        self.type = type

    def info(self):
        print(f"Vegetable Type: {self.type}")

    def calculate(self, values):
        return(max(values))
````

````python
# Implement methods of the two classes in the same manner
for x in [fruit('strawberry'), vegetable('spinach')]:
    x.info()
    print(f'Calculation result: {x.calculate([23, 45, 67])}')
    print()
````
