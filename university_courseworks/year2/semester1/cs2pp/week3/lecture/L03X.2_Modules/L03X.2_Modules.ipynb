<h2 style="color:Crimson;">Using Modules in Python</h2>


**Recall: Python Modular Programming**


- A method of breaking large coding tasks into smaller and more manageable subtasks.

- **Variable** - a name bound to an object in memory, referencing information or references to other objects.

- **Function/Method** - repeatably executable named block of coded instructions, possibly with inputs/outputs.

- **Class** - named encapsulation of related data variables and functions.


- **Module** - a bunch of related code saved in a file with the extension `.py`.
    - `import`able namespace, potentially installable


- **Package** - a directory containing of modules/subpackages. (May have `__init__.py` for setup operations on import.)
    - `import`able namespace, potentially installable 


- **Library** - often used interchangeably with "Package"
    - an umbrella term referring to a reusable chunk of code, a **conceptual** collection of packages and/or modules...perhaps just one of these.
    - Often, this is some code that we can `import` and use.


- **Framework** - describes the basic structured flow and architecture of the application using combinations of the above components, often to assemble applications.


- **Environment** - the collection of code elements that are *visible* to your instance of Python.
    - includes basic Python runtime utilities, additional installed elements, and wider system resources.

### Example

**Break down a larger program into multiple Python modules (files) with *well-defined* functions and interfaces.**

```python
###################################
# file: data_loader.py
def load_csv(filepath): ...
    
###################################
# file: analysis.py
def compute_statistics(data): ...

###################################
# file: main.py or notebook.ipynb

from analysis import compute_statistics

print(compute_statistics(data))
```

- This takes us beyond notebook cell development and single-file scripts into **multi-module** organisation.
- Composing code in this way enables us to implement clear **[separation of concerns](https://www.geeksforgeeks.org/software-engineering/separation-of-concerns-soc/)** and clear project **structure**.
    - Toward [Model-View-Controller](https://www.geeksforgeeks.org/system-design/mvc-architecture-system-design/) architecture
    - Enables more effective, granular testing

Python has a **module system** that allows you to group useful
functions together into a single `.py` file.

- You can `import` Python files and then **access** their functions, classes, and global assignments.


- Importing a `.py` file will **execute all** of the statements in it.
    - If the contents consists solely of **function** and **class** definitions, then those definitions will be made.
    - If other **executable statements** are present, then these will also be executed.


- In Jupyter notebooks, we tend to stick to importing **third-party libraries** to keep the notebook self-contained.
- You can also use `from` to import specific definitions from a module.

<h3 style="color:Crimson;">Aliasing modules</h3>

When **importing** a module, the name of the file acts as an
**identifier** for the contents of the module, and prevents
overwriting of functions and variables using the same names.


Many modules we will be using are often imported
using the `as` keyword. 
- This **aliases** the module and allows you to give it a more concise name.

---

**`moduleSubdirectory/demoModuleSubdirectory.py`** contains the same function and class as the previous example module. 
- It also contains an executable statement, which will be processed on `import`:
```python
print(f'This file is called: {__name__}.')
```

Because its location is directly accessible from the same directory as this notebook, we can `import` its contents as:
```python
import moduleSubdirectory.demoModuleSubdirectory
moduleSubdirectory.demoModuleSubdirectory.kwfunction(...)
```

````python
import moduleSubdirectory.demoModuleSubdirectory

moduleSubdirectory.demoModuleSubdirectory.kwfunction(4, 5, 6)
````

---
This naming structure can easily become overly **verbose**.  

**Aliasing** allows us to reassign the name of the imported module to a name of our choosing.

````python
import moduleSubdirectory.demoModuleSubdirectory as dms

dms.kwfunction(4, 5, 6)
````

<h3 style="color:Crimson;">Selective Imports</h3>

Sometimes, we would like to only import **specific** elements of a module.  We can accomplish this with the `from` statement:

```python

```

These can also be aliased.

````python
from moduleSubdirectory.demoModuleSubdirectory import Student as myst

st = myst()
st.info()
````

<h3 style="color:Crimson;">Import Everything</h3>

Instead of importing a module and using its name or alias as a reference, we can import all (or some) of its contents, retaining their **defined names**.

```python
from moduleSubdirectory.demoModuleSubdirectory import *
```

Then, all of its contents will be directly usable in the local space.

**Caution**: Importing functions and classes in this way can **overwrite** any existing objects with the same name.

````python
from moduleSubdirectory.demoModuleSubdirectory import *

kwfunction(4, 5, 6)
````

<h3 style="color:Crimson;">`__name__` and `__main__`</h3>

When a module is imported or run as a **script** (e.g., from the command line), it is assigned a `__name__`, defined by the system.
- If you are *importing*, the name looks like *the path to the module file*.

- If you are *running* the module as a script, `__name__ = "__main__"`

**`__main__`** is the name of the environment where top-level (entry-point) code  is run.  Even this notebook will have this element defined.

````python
# The __name__ of this Jupyter Notebook

__name__
````

---

**`moduleSubdirectory/demoModuleSubdirectoryScript.py`** contains the construct:
```python
# When running as a script:
if __name__ == "__main__":
    print(f'This file is being run as a script and is called: {__name__}.')
else:
    print(f'This file is being imported and is called: {__name__}.')
```

This allows us to **differentiate behaviour** between the two cases.  

Typically, when `__name__ == "__main__"`, the module can be run as a script, and you will find a function call to begin executing its intended processes.
