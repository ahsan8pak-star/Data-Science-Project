```python
test_codes = [200, 400, 404, 500, 999, "blip"]

for code in test_codes:
    match code:
        case 200:
            print(f"{code} - OK: The request has succeeded.")
        case 400:
            print(f"{code} - Bad Request: The server could not understand the request.")
        case 404:
            print(f"{code} - Not Found: The requested resource could not be found.")
        case 500:
            print(f"{code} - Internal Server Error: The server encountered an error.")
        case _:
            # The underscore is a wildcard and catches any case not listed above
            print(f"{code} - Unknown status code.")
```

```python
test_codes = [200, 400, 404, 500, 999, "blip"]

for code in test_codes:
    match code:
        case 200:
            print(f"{code} - OK: The request has succeeded.")
        case 400 | 404 | 500:
            print(f"{code} - Something unpleasant happened.")
        case 404:
            print(f"{code} - Not Found: The requested resource could not be found.")
        case 500:
            print(f"{code} - Internal Server Error: The server encountered an error.")
        case _:
            # The underscore is a wildcard and catches any case not listed above
            print(f"{code} - Unknown status code.")
```

```python
test_codes = [200, 400, 404, 500, 999, 'blip']
for code in test_codes:
    match code:
        case 200:
            print(f"{code} - OK: The request has succeeded.")
        case 400 | 404 | 500:
            print(f"{code} - Something unpleasant happened.")
        case 404:
            print(f"{code} - Not Found: The requested resource could not be found.")
        case 500:
            print(f"{code} - Internal Server Error: The server encountered an error.")
        case str():
            print(f'{code = } This is a string.')
        case int():
            print(f'{code=} This is an int.')
        case _:
            # The underscore is a wildcard and catches any case not listed above
            print(f"{code} - Unknown status code.")
```

```python
# Let's explore the use of more complex **generic** patterns
def describe(value):
    match value:
        case [x, y]: # matches any sequence of length 2, the brackets are not important
            return f"I found a pair of entries: {x}, {y}."
        case {"name": n, "age": a}: # matches a mapping with those keys
            return f"I found a dictonary entry with 'name' and 'age' keys: name {n}, age {a}."
        case Point(m, n):   # Checks whether the value is a valid instance.
                            # If so, we can access the passed arguments as these local names,
                            # but their values are actually determined by a check of the
                            # class or instance attribute names in the order defined by the
                            # __match_args__ class attribute. It's a messy construct!
            return f"I found a Point instance at ({m}, {n})."
        case _:
            return "Something else"


class Point:
    __match_args__ = ("x", "y")
y = 100
def __init__(self, x, y):
    self.x, self.y = x, y


# Example usage
print(describe( (1, 2) ))
print(describe( {"name": "Alice", "age": 30} ))
print(describe( Point(1, -1) ))

# Test again, changing self.y to be defined as self.z
# or by changing the order of x and y in the __init__.
```

```python
'''
Instead of calling len(nums) twice, the walrus operator assigns the length
to n and simultaneously checks if it is greater than 3.
At the same time, we assign and return 3 to thresh.
'''
nums = [1, 2, 3, 4, 5]
if (n := len(nums)) > (thresh := 3):
    print(f"The list has {n=} elements, which is more than {thresh=}.")
```

```python
'''
For each number n in nums, the expression (square := n * n)
computes and assigns its square to square.
The comprehension then includes only those square values that exceed 10.
'''
nums = [1, 2, 3, 4, 5]
# Compute the square and filter only those squares that are greater than 10.
squares = [square for n in nums if (square := n * n) > 10]
print(squares)
```

```python
"""
Without Walrus Operator (:=), the above operation cannot be efficiently performed with a
comprehension.

Instead the code below requires and relys on a loop, computing the square once per iteration.
"""

result = []
for n in nums:
    square = n * n
    if square > 10:
        result.append(square)
```
