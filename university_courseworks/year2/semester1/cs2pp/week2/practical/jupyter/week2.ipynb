### Slicing

```python
n = int(input("Enter the number of iterations: "))
first_index = int(input("Enter the first index: "))
last_index = int(input("Enter the last index: "))
step = int(input("Enter the step value: "))

for i in range(n):
    slicing_text = input("Enter a text: ")[first_index:last_index:step]
    print(slicing_text)
print("Done!")

```

## Lists

### Lists of Numbers

```python
numbers = [10, 25, 8, 42, 15]
numbers.sort()
min = numbers[0]
max = numbers[-1]
mid = numbers[len(numbers)//2]
mean = sum(numbers) / len(numbers)
lq = numbers[len(numbers) // 4]
uq = numbers[3 * len(numbers) // 4]
iqr = uq - lq
sum = sum(numbers)

num = {
    "num": numbers,
    "sorted": numbers,
    "min": min,
    "max": max,
    "mid": mid,
    "mean": mean,
    "lq": lq,
    "uq": uq,
    "iqr": iqr,
    "sum": sum
}

print(num)
```

### Daffodil Number

```python
n = int(input("Enter a three-digit positive integer: "))

hundreds = n // 100
tens = (n // 10) % 10
ones = n % 10

if hundreds**3 + tens**3 + ones**3 == n:
    print("YES")
else:
    print("NO")
```

### Iterative Problem Solving

```python
n = int(input("Enter a positive integer: "))

sequence = [n]
while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    sequence.append(n)

print(", ".join(str(num) for num in sequence))
```

### Finding the Maximum

```python
numbers = [12, 6, 3, 10, 5, 16, 8, 4, 2, 1]

maximum = numbers[0]
for number in numbers:
    if number > maximum:
        maximum = number

print(maximum)
```

## match/case Statements

### Identifying Collection Patterns

```python
shipments = [
    ["Gadgets", 100],
    ["Widgets", 50, "URGENT"],
    [],
    ["Widgets", 25, "SAFE"],
    ["Gizmo", 10],
    [10, "Gizmo"]
]
```

```python
for shipment in shipments:
    match shipment:
        case [product, quantity]:
            print(f"Processing 2-item shipment: {product} (quantity {quantity}).")
        case [product, quantity, flag]:
            print(f"Found special flag '{flag}': {product} (quantity {quantity}).")
        case _:
            print("Unrecognised shipment pattern.")
```

### Dictionary Patterns

```python
user_records = [
    {"username": "xy993241", "role": "admin", "last_login": "2024-03-01"},
    {"username": "hu183337", "role": "guest"},
    {"username": "oo262911", "last_login": "2024-02-28"},
    {"username": "ne338010", "role": "moderator", "last_login": "2024-01-15"}
]
```

```python
for record in user_records:
    match record:
        case {"role": "admin", "last_login": login}:
            print(f"Praise the admin {record['username']}, last login {login}.")
        case {"role": role}:
            print(f"Greetings, {role} {record['username']}.")
        case _:
            print(f"Unknown role for {record['username']}.")
```

### Dictionary Comprehension 1

```python
dict1 = dict(a=1,
             b=['fish', 'tiger', 'octopus'],
             c=(50, 50),
             d=2,
             e='total',
             f=55,
             g={'chrome': 'tab'},
             h={4, 5, 6})
```

```python
dict2 = dict(c=(50, 60),
             a=12,
             g={'chrome': 'tab'},
             f=55,
             b=['fish', 'tigger', 'octopus'],
             d=2,
             e='total',
             h={4, 5, 6})
```

```python
changed = {key: (dict1[key], dict2[key]) for key in dict1 if dict1[key] != dict2[key]}
print(changed)
```

### Dictionary Comprehension 2

```python
data1 = [12, 82, 35, 82, 12, 35]
data2 = [24, 56, 66, 74, 28, 28]
```

```python
result = {
    key: [data2[i] for i in range(len(data1)) if data1[i] == key]
    for key in sorted(set(data1))
}
print(result)
```

### Euclid, with a walrus

```python
a = 10716
b = 4623

while (remainder := a % b) != 0:
    a, b = b, remainder

print(f"The greatest common divisor is {b}.")
```
