```python
import numpy as np
import pandas as pd
import scipy.stats as stats

# Load dataset
vg_data = pd.read_csv("vgsales.csv").dropna(subset=["Global_Sales"])
original_sales = vg_data["Global_Sales"]

# Define Filter Ranges
# Option A: IQR (25th to 75th percentile) -> Trims 50% of data
q1, q3 = original_sales.quantile(0.25), original_sales.quantile(0.75)
iqr_subset = original_sales[
    (original_sales >= q1) & (original_sales <= q3)
]

# Option B: 10th to 90th Percentile -> Trims 20% of data
p10, p90 = original_sales.quantile(0.10), original_sales.quantile(0.90)
p_10_90_subset = original_sales[
    (original_sales >= p10) & (original_sales <= p90)
]

# Print Statistical Profiles
print(
    f"Original Population -> Count: {len(original_sales):,}, Mean:"
    f" {original_sales.mean():.4f}, Std: {original_sales.std():.4f}"
)
print(
    f"IQR Filter (25th-75th) -> Count: {len(iqr_subset):,}, Mean:"
    f" {iqr_subset.mean():.4f}, Std: {iqr_subset.std():.4f}"
)
print(
    f"10th-90th Percentile -> Count: {len(p_10_90_subset):,}, Mean:"
    f" {p_10_90_subset.mean():.4f}, Std: {p_10_90_subset.std():.4f}"
)

# Kolmogorov-Smirnov Hypothesis Test against Original Distribution
ks_iqr = stats.ks_2samp(iqr_subset, original_sales)
ks_1090 = stats.ks_2samp(p_10_90_subset, original_sales)

print(f"\n--- KS Test Results ---")
print(
    f"IQR Filter vs Original: Statistic = {ks_iqr.statistic:.4f}, p-value ="
    f" {ks_iqr.pvalue:.4e}"
)
print(
    f"10th-90th Filter vs Original: Statistic = {ks_1090.statistic:.4f},"
    f" p-value = {ks_1090.pvalue:.4e}"
)
```
