```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
import joblib

music_data = pd.read_csv('music.csv')
X = music_data.drop(columns = ['genre'])
y = music_data['genre']

model = DecisionTreeClassifier()
model.fit(X, y)

model = joblib.dump(model, 'music-recommender.joblib') # file already existed

model = joblib.load('music-recommender.joblib')
predictions = model.predict([[21, 1], [22, 0]]) # 21 year old Male & 22 year old Female respectively
predictions
```
