```python
import numpy as np
import pandas as pd
from data_outlier import OutlierCapper

# Initialise the transformer with standard 1.5 IQR factor
capper = OutlierCapper(factor = 1.5)
```

```python
# Standard list with upper (100) and lower (-50) outliers
sample_with_outliers = [10, 12, 11, 13, 12, 11, 14, 12, 100, -50]
```

```python
# Multi-column DataFrame with extreme outliers in 'age' and 'income'
sample_df = pd.DataFrame({
    "age": [25, 28, 30, 29, 27, 140, 26],         # 140 is an age outlier
    "income": [50000, 52000, 48000, 51000, 49000, 53000, 1500000] # 1,500,000 is an income outlier
})
```

```python
# Convert 1D list into a 2D DataFrame for OutlierCapper
df_outliers = pd.DataFrame({"original_value": sample_with_outliers})
```

```python
# Apply OutlierCapper
df_capped = capper.fit_transform(df_outliers)
```

```python
# Combine into a single visual comparison table
comparison_1d = pd.DataFrame({
    "Original": df_outliers["original_value"],
    "Capped": df_capped["original_value"]
})
```

```python
print("=== 1D Outlier Capping Visual Output ===")
display(comparison_1d)
```

```python
# Apply OutlierCapper across numeric columns
capped_df = capper.fit_transform(sample_df)
```

```python
print("=== Original DataFrame ===")
display(sample_df)
```

```python
print("\n=== Capped DataFrame (Outliers Trimmed to IQR Bounds) ===")
display(capped_df)
```

```python
# Replaces test_data_outliers.py assertions directly inside the notebook
test_df = pd.DataFrame({"value": [1, 2, 3, 4, 100]})
transformed_test = capper.fit_transform(test_df)
```

```python
# Visual check & Assertion
max_val = transformed_test["value"].max()
assert max_val < 100, f"Assertion Failed: Max value is {max_val}"
```
